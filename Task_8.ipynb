```python
import google.generativeai as genai
import os, json
from IPython.display import display, Markdown

genai.configure(api_key=os.environ['GEMINI_API_KEY'])
MODEL = 'gemini-2.5-flash'

def ask(messages, max_tokens=1500, system=None):
    """
    Unified ask() that mirrors the Anthropic interface.
    messages: list of {'role': 'user'|'assistant', 'content': str}
    system:   optional system prompt string
    """
    model = genai.GenerativeModel(
        model_name=MODEL,
        system_instruction=system or ''
    )
    # Convert messages list → Gemini history + final prompt
    history = []
    for msg in messages[:-1]:
        role = 'user' if msg['role'] == 'user' else 'model'
        history.append({'role': role, 'parts': [msg['content']]})
    final_prompt = messages[-1]['content']
    chat = model.start_chat(history=history)
    response = chat.send_message(
        final_prompt,
        generation_config=genai.GenerationConfig(max_output_tokens=max_tokens)
    )
    return response.text

print(f'Gemini client ready. Model: {MODEL}')
```

### Step 2: 4 Prompting Strategies

| # | Strategy | Intent |
|---|---|---|
| P1 | Simple Summarise | Baseline — no guidance |
| P2 | Contrastive Analysis | Force gap detection between companies |
| P3 | Chain of Thought | Step-by-step adversarial reasoning |
| P4 | Role-Based (SEC Attorney) | Adversarial framing for legal omissions |

```python
prompts = {
    'P1_Simple': {
        'label': 'Prompt 1 — Simple Summarise',
        'messages': [{'role':'user','content':
            f'Summarize the key risks mentioned in these two 10-K filings.\n\n{CONTEXT}'
        }]
    },
    'P2_Contrastive': {
        'label': 'Prompt 2 — Contrastive Analysis',
        'messages': [{'role':'user','content':
            f'''You are a financial analyst. Compare the Risk Factors of Block vs PayPal 2023 10-Ks.\n
Answer:\n
1. What risks does Block disclose that PayPal does NOT mention?\n
2. What risks does PayPal disclose that Block does NOT mention?\n
3. Which company uses more hedged or vague risk language?\n\n{CONTEXT}'''
        }]
    },
    'P3_ChainOfThought': {
        'label': 'Prompt 3 — Chain of Thought',
        'messages': [{'role':'user','content':
            f'''Think step by step as a senior risk analyst at a hedge fund.\n
Step 1: Identify every risk category Block mentions (AI, Bitcoin, regulatory, operational, legal).\n
Step 2: For each, assess whether Block disclosure language is specific or evasive.\n
Step 3: Compare to PayPal — does PayPal use more conservative, specific language for similar risks?\n
Step 4: Flag Block risks that appear underweighted or buried.\n
Step 5: Write your final risk divergence verdict.\n\n{CONTEXT}'''
        }]
    },
    'P4_RoleBased': {
        'label': 'Prompt 4 — Role-Based (SEC Attorney)',
        'messages': [{'role':'user','content':
            f'''You are an SEC enforcement attorney reviewing these 10-K filings for disclosure violations.\n
Find:\n
- Bitcoin/crypto risks inadequately disclosed given Block\'s deep exposure\n
- AI operational risks mentioned generically but lacking specificity\n
- Language designed to technically disclose a risk while minimising investor alarm\n
- Differences in how Block vs PayPal handle regulatory uncertainty\n
Flag specific phrases that could become the basis of an SEC comment letter or investor lawsuit.\n\n{CONTEXT}'''
        }]
    },
}

results = {}
for key, p in prompts.items():
    print(f'Running {key}...')
    results[key] = {'label': p['label'], 'response': ask(p['messages'])}
    print(f'  ✓ {len(results[key]["response"]):,} chars')

print('\nAll 4 prompts complete.')
```

```python
for key, r in results.items():
    display(Markdown(f'### {r["label"]}\n\n{r["response"]}\n\n---'))
```

### Step 3: LLM as Judge

```python
judge_input = '\n\n'.join(
    f'=== {r["label"]} ===\n{r["response"]}'
    for r in results.values()
)

judge_prompt = f'''You are an expert evaluator assessing the quality of 4 different LLM-generated
risk analyses of Block vs PayPal 2023 10-K filings.

Rate each analysis (P1–P4) on a scale of 1–10 for each criterion:
- DEPTH: How thoroughly does it explore the risk landscape?
- SPECIFICITY: Does it cite actual risk language vs vague generalities?
- ACTIONABILITY: Could an investor or analyst act on these insights?
- RISK DETECTION: Did it uncover subtle/undisclosed risks vs surface-level ones?

Then rank P1–P4 from best to worst with a one-sentence rationale for each.
Finally, identify which single insight across ALL outputs is most valuable.

Here are the 4 analyses:\n\n{judge_input}'''

judge_verdict = ask(
    [{'role':'user','content': judge_prompt}],
    max_tokens=1500,
    system='You are a rigorous, impartial evaluator of financial risk analysis quality. Be critical and specific.'
)

display(Markdown('### Judge Verdict\n\n' + judge_verdict))
```
